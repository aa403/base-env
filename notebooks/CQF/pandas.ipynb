# Pandas

Another Python flagship library, Pandas is named after 'panel data'. 

It is a package of fast, easy, efficient data analysis tools for Python. Its popularity has surged in recent years, coincident with the rise of fields such as data science and machine learning. Just as the binomial model has enabled (non-mathematical) Finance, MBA and CFA students to do option pricing, Pandas allows non-computer scientists to analyse and work effectively with data (away from Excel). In fact that was a main reason why its creater, Wes McKinney, developed the library. <br>
We are all familiar with a spreadsheet. It is an interactive and powerful Python computer application for organising, analysing and storing data in tabular form. More generally it can be referred to as a special case of a **DataFrame**. That is, a 2-dimensional labelled data structure with columns of potentially different types. It accepts many different types of input. <br>
Pandas provides high-performance data structures and is designed to make data analysis fast and easy in Python. It is built on top of NumPy. There are two main structures (or classes) in Pandas:

1. Series - 1D labelled homogeneous array, size immutable.
2. Data Frames - 2D labelled, size-mutable tabular structure with potentially heterogeneously typed columns.

In practice, the DataFrame is much more useful since it includes useful information such as column names read from the data source. <br>


## Series
A Series contains a one-dimensional array of data, and an associated sequence of labels called the index; and are the equivalent of one-dimensional arrays. As defined above DataFrames are two-dimensional and are collections of Series. A Series is initialized using a list, tuple, or a NumPy array. The simplest Series is formed from only an array of data. The axis labels are collectively called index.

```python
#from pandas import * # or
import pandas as pd # preferable
import math as m
import numpy as np
#import matplotlib.pyplot as plt
```

The simplest Series is formed from only an array of data. To create an empty series we can do so by

```python
# Create an Empty Series
s = pd.Series(dtype=float)
print(s)
s
```

This can be populated using the arange() function

```python
# Create a Series from ndarray
s = np.arange(10,20)
print((s))
pd.Series(s)
```

Creating a series from a list and tuple

```python
k1=[2,3,4,5,6]; k2=(2,3,4,5,6)
s1=pd.Series(k1); s2=pd.Series(k2)
print((type(s1))) ; print(type(s2))
```

working backwards

```python
y=list(s1)
type(y)
```

```python
series1=pd.Series([ 1 , 4 , 9 , 16 ]); # could have from pandas import Series
print(series1)
print(type(series1))
# prints the following
```

There are similarities and differences between series and arrays. **Series, like arrays, are sliceable**. However, unlike a 1-dimensional array, a Series has an additional column,  an index, which is a set of values associated with the rows of the Series. In the example above, the programmer hasn't provided any indexing. Pandas has automatically generated an index by default using the sequence 0, 1, . . . . It is also possible to use other values as the index when initializing the Series using a keyword argument. If we want to use our own then the earler code (and output) becomes

```python
series1=pd.Series( [ 1 , 4 , 9 , 16 ], index=[ 'x' , 'y' , 'z' , 't' ] )
print(series1)
```

More efficient to pass through lists in defining a Series

```python
list1=[ 1 , 4 , 9 , 16 ]
index=[ 'x' , 'y' , 'z' , 't' ]
series2=pd.Series(list1, index )
series2
print(type(index))
```

Compared with a regular NumPy array, you can use values in the index when selecting single values or a set of values

```python
series1["z"]
```

```python
series1[2] # using the numeric index outputs the same
```

A Series is mutable

```python
series1["t"]= -49 # element at index t is assigned the value 49
series1 # series are compound types
```

NumPy array operations, such as filtering with a boolean array, scalar multiplication, or applying maths functions, will preserve the index-value link

```python
new_series=3*series1
new_series
```

```python
series1[series1 > 0] # can also use print
```

Now create a new series2 from series1

```python
series2=series1[series1>0]
series2 
```

```python
import numpy as np # maths functions in numpy
np.exp(series1)
```
