```python
df = pd.read_html('https://en.wikipedia.org/wiki/House_of_Commons_of_the_United_Kingdom')[2]
# df
```

```python
df.columns = df.columns.to_flat_index()

df = df[df.columns[1:]]

df.columns = ["party", "members_2019", "members_2024", "delta"]

df.loc[1].party = "Labour"
df.loc[12].party = "Reform"

df.loc[8].members_2019 = 0
df.loc[8].members_2024 = 2

df.loc[11].members_2019 = 0
df.loc[11].members_2024 = 1

df.loc[14].members_2019 = 0
df.loc[14].members_2024 = 17

df.drop(index=[16, 17, 18], inplace=True)

df.members_2024 = df.members_2024.apply(lambda x: int(x))
df.members_2019 = df.members_2019.apply(lambda x: int(x))
                      
df["delta"] = df.members_2024 - df.members_2019
df.set_index("party", inplace=True)
df
```
