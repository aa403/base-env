```python
def litres_to_pints(l: float) -> float:
    return round(l * 1.75975, 4)


def litres_to_gallons(l: float, kind="Imperial") -> float:

    conversions = {"Imperial": 0.219969, "US": 0.264172}

    match kind:
        case "Imperial" | "US":
            return round(l * conversions[kind], 4)

        case _:
            raise ValueError(f"kind must be Imperial or US, {kind} provided")
```

```python
litres_to_pints(1)
```

```python
litres_to_gallons(4, kind="US")
```

```python
litres_to_gallons(4, kind="Imperial")
```

```python
def kg_to_stone_and_lbs(kg) -> (float, float):
    total_lbs = 2.2 * kg
    lbs, stones = m.modf(total_lbs / 14)
    lbs *= 14
    return stones, lbs, total_lbs
```

```python
kg_to_stone_and_lbs(72)
```

```python
def height_to_feet_and_inches(cm: float) -> (float, float):
    total_inches = cm / 2.54

    remaining_inches, feet = m.modf(total_inches / 12)
    remaining_inches *= 12
    # remaining_inches = (total_inches % feet)

    return feet, remaining_inches, total_inches
```

```python
height_to_feet_and_inches(181)
```

```python

risk_free_rate = 0.036

def disc_compunding(amount, rate):
    for i in range(3):
        r = amount * (1 + rate)**(i+1)
        yield round(r, 2)
```

```python
dc = disc_compunding(100, risk_free_rate)
[_ for _ in dc]
```

```python
def cont_compounding(amount, rate):
    for i in range(3):
        r = amount * m.exp(rate * (i+1))
        yield round(r, 2)
```

```python
dc = cont_compounding(100, risk_free_rate)
[_ for _ in dc]
```

```python
def d_discount(fv, rate, periods):
    return fv / (1 + rate)**periods


def c_discount(fv, rate, periods):
    return fv / m.exp(rate * periods)
```

```python
d_discount(5000, 0.06, 3)
```

```python
c_discount(5000, 0.06, 3)
```
